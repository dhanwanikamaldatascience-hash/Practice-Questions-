```python
# Section c =Question 1

def prime(n):
    i=2
    while(i<n):
        if(n%i==0):
            print("not a prime no ")
            break
        i=i+1
    else:
            print("prime no ")


n=int(input("Enter a non to check prime of not "))
prime(n)
```

```python
# Section c =Question 3


def fib(n):
    if n==0 or n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)
n=int(input("enter n no of element "))  

for i in range(n):
    print(fib(i))
```

```python
# Section c =Question 4

def palindrome(n):
    if n == n[::-1]:     # Reverse check
        return True
    else:
        return False

n= input("Enter a string: ")

if palindrome(n):
    print("Palindrome")
else:
    print("Not a palindrome")
```

```python
# Section c =Question 6
def count_vo_co(n):
    vowels="aeiou"
    vcount=0
    ccount=0

    for i in n:
        if i in vowels:
            vcount=vcount+1


    for i in n:
        if i.isalpha() and i not in vowels:
            ccount=ccount+1

    return vcount,ccount

n=input("enter a string ")
v,c=count_vo_co(n)
print("vowels",v)
print("constant",c)
```

```python

# Section c =Question 7
def dup(lst):
    new_list = []
    for i in lst:
        if i not in new_list:
            new_list.append(i)
    return new_list

lst = [21, 15, 16, 19, 20, 10, 7, 15, 19]
a = dup(lst)
print(a)
```

```python
# Section d =Question 5
class Animal:
    def __init__(self,legs,eyes,state):
        self.legs=legs
        self.eyes=eyes
        self.state=state

    def display(self):
        print(f"the legs of this animal is {self.legs} and eyes is {self.eyes} and state is {self.state}")


class Dog(Animal):
    def __init__(self,legs,eyes,state,tail,speak):
        super(). __init__(legs,eyes,state)
        self.tail=tail
        self.speak=speak



    def display(self):
        print(f"dogs have {self.legs} legs and {self.eyes} eyes and {self.state} and they speak{self.speak}")


a=Dog(4,2,"road",1,"bark")
a.display()



```
